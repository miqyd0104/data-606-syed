# Optional Pandas preparation
Illustrative sales records. A Series is a labelled 1D collection; a DataFrame is a labelled table. This is preparation, rather than proof of completing a later assessment.

```python
import pandas as pd
import io
frame = pd.DataFrame({'store': ['A', 'B', 'C'], 'sales': [100.0, 150.0, None]})
print(frame)
print('Shape:', frame.shape)
print('Columns:', list(frame.columns))
print('Head:', frame.head(), sep='\n')
frame.info()
print(frame.describe())
```

## Select, add and summarise
A single column is a Series. loc selects labels/conditions; iloc selects positions.

```python
print('Sales column:', frame['sales'], sep='\n')
print('First two rows:', frame.iloc[:2], sep='\n')
print('Store B:', frame.loc[frame['store'] == 'B'], sep='\n')
frame['region'] = ['North', 'South', 'North']
frame.loc[len(frame)] = ['D', 200.0, 'South']
print('Updated table:', frame, sep='\n')
print('Observed mean sales:', frame['sales'].mean())
assert frame.shape == (4, 3)
```

## CSV and missing values
Read an illustrative CSV through a text stream. Missing sales are excluded from this analysis instead of being silently treated as zero.

```python
csv_text = "store,sales\nA,100\nB,150\nC,\n"
loaded = pd.read_csv(io.StringIO(csv_text))
print('Missing counts:', loaded.isna().sum(), sep='\n')
cleaned = loaded.dropna(subset=['sales'])
print('Cleaned:', cleaned, sep='\n')
print('Total known sales:', cleaned['sales'].sum())
assert len(cleaned) == 2
assert cleaned['sales'].sum() == 250
```
