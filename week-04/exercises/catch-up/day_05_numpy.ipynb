## 1. Required shapes
A 1D array has 12 values; the 2D array has 3 rows and 4 columns; the 3D array has 2 blocks, each with 5 rows and 4 columns.

```python
one_d = np.arange(1, 13)
two_d = np.arange(1, 13).reshape(3, 4)
three_d = np.arange(1, 41).reshape(2, 5, 4)
for name, array in [('1D', one_d), ('2D', two_d), ('3D', three_d)]:
    print(name, 'shape:', array.shape, 'dimensions:', array.ndim, 'size:', array.size)
    print(array)
assert one_d.shape == (12,)
assert two_d.shape == (3, 4)
assert three_d.shape == (2, 5, 4)
```

## 2. Student scores
Find the highest, lowest, mean and total. Adding 5 to an array adds 5 to each value. No score cap was specified.

```python
scores = np.array([65, 72, 88, 91, 54, 80])
print('Highest:', scores.max(), 'Lowest:', scores.min())
print('Average:', scores.mean(), 'Total:', scores.sum())
bonus_scores = scores + 5
print('With bonus:', bonus_scores)
assert scores.sum() == 450
assert np.isclose(scores.mean(), 75)
assert np.array_equal(bonus_scores - scores, np.full(scores.shape, 5))
```

## 3. Weather
Boolean indexing selects matching temperatures. Strict comparisons exclude 25 from above 25 and 20 from below 20.

```python
temperatures = np.array([18, 20, 22, 26, 29, 19, 25])
hot = temperatures[temperatures > 25]
cool = temperatures[temperatures < 20]
print('Above 25°C:', hot, 'Count:', hot.size)
print('Below 20°C:', cool, 'Count:', cool.size)
print('Average temperature:', round(float(temperatures.mean()), 2))
assert hot.tolist() == [26, 29]
assert cool.tolist() == [18, 19]
```

## 4. Retail sales
Rows are stores; columns are weeks. Reducing axis 1 gives a result per store; reducing axis 0 gives a result per week.

```python
sales = np.array([[100, 120, 110, 130], [90, 105, 115, 100], [140, 135, 150, 145]])
print('Sales:', sales, sep='\n')
print('Total per store:', sales.sum(axis=1))
print('Average per store:', sales.mean(axis=1))
print('Total per week:', sales.sum(axis=0))
print('Average per week:', sales.mean(axis=0))
print('Grand total:', sales.sum())
assert sales.sum(axis=1).tolist() == [460, 410, 570]
assert sales.sum(axis=0).tolist() == [330, 360, 375, 375]
```

## 5. Customer purchases
Use a fixed seed so this example is repeatable. Ten customers is an illustrative choice. The upper bound is exclusive, so 501 allows £500.

```python
rng = np.random.default_rng(606)
purchases = rng.integers(1, 501, size=10)
print('Purchase amounts (£):', purchases)
print('Total (£):', purchases.sum(), 'Average (£):', purchases.mean())
assert purchases.shape == (10,)
assert np.all((purchases >= 1) & (purchases <= 500))
```

## 6. Reshape and transpose
Reshape reorganises the same number of values into a new shape. Transpose swaps the axes; for 2D data, columns become rows.

```python
original = np.arange(1, 7).reshape(2, 3)
reshaped = original.reshape(3, 2)
transposed = original.T
print('Original:', original, sep='\n')
print('Reshaped:', reshaped, sep='\n')
print('Transposed:', transposed, sep='\n')
assert reshaped.shape == transposed.shape == (3, 2)
assert not np.array_equal(reshaped, transposed)
```

## 7. Indexing, iteration and saving
Indexing starts at zero. Iteration over a 2D array produces rows. A temporary folder demonstrates binary and CSV round trips without leaving files behind.

```python
from pathlib import Path
from tempfile import TemporaryDirectory
print('First value:', two_d[0, 0])
print('First row:', two_d[0])
print('Second column:', two_d[:, 1])
for row in two_d:
    print('Row total:', row.sum())
with TemporaryDirectory() as directory:
    binary = Path(directory) / 'array.npy'
    text = Path(directory) / 'array.csv'
    np.save(binary, two_d)
    np.savetxt(text, two_d, delimiter=',', fmt='%d')
    assert np.array_equal(np.load(binary), two_d)
    assert np.array_equal(np.loadtxt(text, delimiter=',', dtype=int), two_d)
print('Binary and CSV round trips passed')
```
